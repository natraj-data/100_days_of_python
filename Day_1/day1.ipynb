```python
# Variables
# We have 2 variables glass1 and glass2. glass1 contains milk and glass2 contains juice. 
# Write 3 lines of code to switch the contents of the variables. 
# You are not allowed to type the words "milk" or "juice".
#  You are only allowed to use variables to solve this exercise.

# way 1 
glass1 = "milk"
glass2 = "juice"

temp = glass1
glass1 = glass2
glass2 = temp
print(glass1, glass2)

# way 2 
glass1 = "milk"
glass2 = "juice"
glass1, glass2 = glass2 , glass1
print(glass1, glass2)

# Variable naming 
# it has to be more self explanatory and descriptive in nature to improve the code readable.
# we can use the multiple words in variable name with the splitters like _ to split the words. 
# Shouldn't have the space as a splitter
# avoid using the reserved words as a variable name. 
# The variables names shouldn't start with number or any symbols than $. 
```
