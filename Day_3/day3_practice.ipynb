```python
# BMI Calculator with interpretation 
# Add some if/elif/else statements to the BMI calculator so that it interprets the BMI values calculated.
# If the bmi is under 18.5 (not including), print out "underweight"
# If the bmi is between 18.5 (including) and 25 (not including), print out "normal weight"
# If the bmi is 25 (including) or over, print out "overweight"

height = float(input("Enter your height in cm :\n"))
weight = float(input("Enter your weight in kg: \n"))

bmi = weight / ((height/100)**2) 
print(bmi)

if bmi < 18.5:
  print("underweight")
elif 18.5 <= bmi <25:
  print("normal weight")
elif bmi >=25:
  print("overweight")
```

```python
# Multiple if statement in succession 
bill = 0
height = int(input("Whats your height in cm ? \n"))
if height >=120:
  print("You can ride the roller coaster")
  age = int(input("What's your age in integer? \n"))
  if age < 12:
    bill += 5 
  elif 12 <= age <= 18:
    bill += 7
  else:
    bill += 12
  
  wants_photo = input("Do you wanna take the picture in the ride: y or n").lower()
  if wants_photo == "y":
    bill += 3
  
  print(f"Your final bill is ${bill}")
else:
  print("You have to grow taller to ride..")
```
