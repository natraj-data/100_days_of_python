```python
# Take three side lengths. First, check if they can actually form a triangle (the sum of any two sides must be greater than the third). If valid, classify it as Equilateral, Isosceles, or Scalene.

def triangular_classifier (x:float, y:float, z:float)-> str:
  # Guard clause
  if (x<=0) or (y<=0) or (z<=0):
    return "Invalid side: Sides must be greater than 0"
  # Triangular inequality theorem : sum of any two sides has to be greater than the other
  if (x + y <= z) or (y + z <= x) or (x + z <= y) :
    return "Invalid triangle: These sides cannot make the closed triangle"
  # classify the valid triangle
  if (x == y == z) :
    return "Equilateral Triangle"
  elif (x == y) or (y == z) or (x == z):
    return "Isosceles Triangle"
  else:
    return "Scalene Triangle"

print(triangular_classifier(-2,4,6))
print(triangular_classifier(1,2,3))
print(triangular_classifier(5,5,5))
print(triangular_classifier(5,5,6))
print(triangular_classifier(4,5,7))
```

```python
# 3. Smart Thermostat Logic. Given current_temp, target_temp, is_home (boolean), and eco_mode (boolean), decide whether to turn on the "Heater", "AC", or "Off". If is_home is False, allow a 5-degree buffer before turning anything on; if eco_mode is True, add an extra 2-degree buffer.

def control_thermostat(current_temp, target_temp, is_home, eco_mode):
  # 1. Calculate the allowed temperature buffer
  buffer = 0
  if not is_home:
    buffer += 5
  if eco_mode:
    buffer += 2

  # 2. Decide the thermostat action based on the buffer
  if current_temp < target_temp - buffer:
    return "Heater"
  elif current_temp > target_temp + buffer:
    return "AC"
  else:
    return "Off"


# --- Example Usage ---
print(
    control_thermostat(current_temp=68, target_temp=72, is_home=True, eco_mode=False)
)  # Output: Heater
print(
    control_thermostat(current_temp=68, target_temp=72, is_home=False, eco_mode=False)
)  # Output: Off (68 is within the 5-degree buffer of 72)
print(
    control_thermostat(current_temp=80, target_temp=72, is_home=False, eco_mode=True)
)  # Output: AC (80 exceeds 72 + 7 buffer)
```
