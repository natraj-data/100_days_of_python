```python
# Primitive data types

# Data is in distinct form that has to be recognized by the python. Python has 4 primitive basic data types.
# str, int, float, bool 

# str -> "Hello" any textual data wrapped up by the quotes are considered as a str

# int -> any whole number considered as a integer int 
# float -> any number has the decimal places, then it could be considered as a float value, float.
# bool -> True/False these two possibilities in the bool data types. 
```

```python
# str: 
# str -> "Hello" any textual data wrapped up by the quotes are considered as a str
print("Hello"[0]) # subscripting 
print("Hello"[1]) # pull out the second letter of the given string
print("Hello"[-1]) # to pull out the last character in the string 
print("123" + "345") # concat 

fname = "natarajan"
lname = "thanaraj"
fullname = fname + " " +lname 
# "123" is str , but not int because it is wrapped up by the quotes 
print(fullname)

# length of the str value,  len(variable)
print("Length of the fullname is : " , str(len(fullname)))

# To check the datatype of the variable, type(variable) 
print("Type of the fname : ", fname , "is " , type(fname))

# subscripting to access the any element in the str value using their index/position. 
print(fname[0]) # to grab the 1st letter of fname
print(fname[1]) # to grab the 2nd letter of fname
print(fname[-1]) # to grab the last letter of fname

# String concatenation 
print("123" +"345") # it concat both the str

```

```python
# number datatypes are int , float 
# Integer -> whole number 
print(123 + 345) 

# large numbers 
print(123_345_567) # _ to separate the digits that could ease humans to understand its intensity of values. 

# float-> floating point number -> means that the number has the decimal point float around the number 
print(3.14159)
print(type(3.14159))

# Boolean (True/ False)
print(True)
print(type(True))
print(type(False))
```

```python
# Type error, Type checking and Type conversion 
# wkt, when we pass the wrong datatype to the function like len(), it throws the type error before
# lets get to know what is type error. Before that we need to know about the function a little bit to understand it better how it really means the type error we got. 

# Imagine a function as a machine in a factory. potatoes into chips (Peeling -> cutting -> frying)
# What if we pass the rock instead of potatoes into the same machine, machine could leave a warning hey, something coming up instead of potatoes. So function generally expect the right argument/ datatype that get recognized by the function, If we fed something different that it doesn't recognized by the function, then it leaves the type error. Thats how we end up with the type error when we pass the integer into the len function before. 

# The len function generally doesn't work with integer, by forcing this through we end up with an error and our code breaks and it gives is the type error. 

# By definition of len function, we could use this function to measure the no of character/ length of elements for the sized data types. Here the arguments are sized datatypes/ sequences are string, bytes, tuple, list, or range or a collections such as dictionary ,set , frozen set. 

print(len(str(123456))) # turn the integer into str by str() and check its length/ how many digits. 
print(type("Hello")) # <class 'str'>
print(type(123)) # <class int>
print(type(123.45)) # <class float>
print(type(True)) # <class bool>

print(type(int("123")))
# print(type(int("abc"))) # value error
# ValueError: invalid literal for int() with base 10: 'abc'
print(int("123") + int(345))
```

```python
# Type conversion using the str(), int() , float() , bool()
# str()
print(str(123), type(str(123)))
print(str(True), type(str(True)))
print(str(123.45),type(str(123.45)))
print(str([1,2,3,4,5]),type(str([1,2,3,4,5])))
print(str({1,2,3,4,5}), type(str({1,2,3,4,5})))
print(str({"name": "Natraj" , "age": 32}), type(str({"name": "Natraj" , "age": 32})))
print(str(("name", "natarajan")), type(str(("name", "natarajan"))))

# int()
print(int("123"), type(int("123")))
# print(int("abc"), type(int("abc"))) # it returns the value error 
print(int(True), type(int(True)))
print(int(False), type(int(False)))

# print(int(None), type(int(None))) #TypeError: int() argument must be a string, a bytes-like object or a real number, not 'NoneType'

print(int(123.56) , type(int(123.45))) 
# print(int([1,2])) int() argument must be a string, a bytes-like object or a real number, not 'list'

print(int(b'0121'), type(int(b'0121')))
print(type(b'0121'))

# Bool
print(bool("123"), type(bool("123")))
print(bool(123), type(bool(123)))
print(bool(12.45), type(bool(12.45)))
print(bool("abc"), type(bool("abc")))
print(bool([1,2,3,4]), type(bool([1,2,3,4])))
print(bool((1,2,3)), type(bool((1,2,3))))
print(bool({1,2,3,4}), type(bool({1,2,3,4})))
print(bool({"a":1}), type(bool({"a":1})))
print(bool(0), type(bool(0)))
print(bool(""), type(bool("")))
print(bool({}), type(bool({})))
print(bool([]), type(bool([])))
print(bool(()), type(bool(())))
print(bool(0.0), type(bool(0.0)))

# challenge
print("Number of letters in your name: " + str(len(input("What's your name:\n"))))
```

```python
# Mathematical operations 
# int (operator) int -> int mostly but the division returns the float 
# Here while division, the implicit conversion happens so that the result becomes as float instead of the int. 
print(1 + 1) # add
print(1-1) # subtract
print(1*1) # multiply 
print(3/2) # divide -> return the float, implicit casting happens 
print(3//2) # floor division -> returns the int, truncate the decimal places with the point. implicit happens
print(3**2) # exponent, returns int 

print(" Other blocks ".center(50, "*"))

# When one of the operand is float and the other is int -> then you can expect all the result return the float values. So here the type casting happens implicitly. 
print(1 + 1.1) 
print(1-1.1)
print(1*1.1)
print(3/2.5) 
print(3//2.5)
print(3**2.5)

print(" Other blocks ".center(50, "*"))
# When both the operands are float, how about the result 
print(1.0 + 1.1) 
print(1.3-1.1)
print(1.2*1.1)
print(5.0/2.5) 
print(3.25//2.5)
print(3.5**2.5)

# When there is an expression includes all the operation with/without different level. 
# Then how could we deal such expression and evaluate it. 
# In python, the operation goes from left to right when there is no priority/ precedence constructor like (). 
# If any precedence, them it follows the some different order. LEFT TO RIGHT , PEMDASLR, MD ARE SAME LEVEL.


# Order of operation PEMDAS 
# parans, exponent, multiplication/division, addition/subtraction

# ()
# **
# * or /
# + or -

# Whats is the output of the code 
print(3 * 3 + 3 / 3 -3) # 7.0 

# Change the code so that outputs 3 
print(3 * (3 + 3) / 3 -3) # 3.0
```

```python
# BMI Calculator 
# The body mass index (BMI) is a measure used in medicine to see if someone is underweight or overweight. This is the formula used to calculate it:
# bmi is equal to the person's weight divided by the person's height squared.
# Convert this sentence into code on line 6.

# BMI = (Wt in kg) / (Ht in m) (MKS)
# BMI = (Wt in lbs) / (Ht in inch) (UCS)
print("Welcome to BMI Calculator")
weight = float(input("Enter the weight of your's in kg: \n"))
height = float(input("Enter the height of your's in cm: \n"))
bmi = (weight)/((height/100)**2)
print(f"Your BMI is : {round(bmi)}")
```

```python
# Assignment Operator 

# += , -= , *= , /= , //= , **= 

score = 2
score += 5
print(score)

score -= 2
print(score)

score *= 3
print(score)
```

```python
# f-String -> help us to create more flexible string creation where we can incorporate the different datatype-ed values or we can manipulate the values withing the string creation easily. 

score = 0 
height = 1.8 
is_winning = True 

print("Your score is " + str(score) + "." + "Your height is " + str(height) + ". And your winning is " + str(is_winning))

print(f"Your score is {score}, height is {height} , and your winning is {is_winning}")
```
