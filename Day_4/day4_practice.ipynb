```python
import random as r

# Random function for the random number , integer and floating point number
 
# To generate the random value from (included) 0.0 to 1.0(excluded)
print(r.random()) 

# To generate the random integer within the given range of values (both sides were included)
print(r.randint(1,10)) # return N integer => a <= N <= b (both end points are included)

# To generate the random floating point number (a <= float value <= b) both end are included
print(r.uniform(1,10))

# To generate the randomly selected element from the given range (start, stop, step), start included but not the stop
print(r.randrange(1,10,2)) # this return any one of the element from the range (1,3,5,7,9) randomly. 
```

```python
# sequence selection and shuffle 

alphabets = [i for i in "abcdefghijklmnopqrstuvwxyz" ]
print(alphabets)

numbers = [i for i in range(0,10)]
print(numbers)

# Lets import the random module and use it for the sequence selection and sequence shuffling 
import random as r
print(r.choice(alphabets)) # returns any one of the alphabet from the list 
print(r.choices(alphabets, k = 4)) # it return the subset of k random elements with the repeat elements
print(r.sample(alphabets,10)) # sample of alphabets which contains unique k items 
clone_alphabets = alphabets.copy() # it create the copy of the alphabets 
r.shuffle(clone_alphabets) # this shuffle the original list, returns None 
print("alphabets", alphabets)
print("shuffled_alphabets" , clone_alphabets)
```

```python
# random module -> seed , getstate , setstate
import random as r 

# 1. Initialize with a fixed seed for reproducibility
r.seed(42) 
print(r.random()) 

# 2. Save the current internal state
saved_state = r.getstate()

# 3. Generate a few numbers
print(r.randint(1, 100))  # 4
print(r.randint(1, 100))  # 95

# 4. Restore the saved state to rewind the generator
r.setstate(saved_state)

# 5. Calling the functions again produces the exact same results
print(r.randint(1, 100))  
print(r.randint(1, 100)) 

# In data science, managing the state of Python's random module ensures reproducibility, fault tolerance, and fair experimentation across stochastic (random) workflows. While modern libraries like NumPy and PyTorch have their own state managers, Python's built-in random module is still widely used in data pipelines, custom algorithms, and pure-Python simulations.

import random

def simulate_demand(days=5):
    return [random.randint(50, 150) for _ in range(days)]

random.seed(101)

# Capture the state before testing Strategy A
baseline_state = random.getstate()
demand_for_strategy_a = simulate_demand()

# Rewind the state so Strategy B faces the exact same customer demand
random.setstate(baseline_state)
demand_for_strategy_b = simulate_demand()

print("Strategy A Demand:", demand_for_strategy_a)  # [124, 74, 119, 95, 109]
print("Strategy B Demand:", demand_for_strategy_b)  # [124, 74, 119, 95, 109]
```

```python
# Copy the list 
# Method 1 : just assign the variable as a value into other variable. But it shares the reference object means each variable point the original object,. so if we make any change in the object by any variable, it could reflect in the other as well. 
names = ["Natraj" , "Leela" , "Thanaraj" , "Kariaklan"]
family_names = names
names.append("Jeyalakshmi")
print(family_names)

# Method 2 : shallow copy 
# It resolves the referential error in the last method. But it works only for the 1d list. If we have multi dimensional list, it still hold connection with the original list in some form.
sales = [21, 34 , 45 ,65, 76, [12,34,43,44] ]
sales_copy = sales.copy()
sales.insert(2, 25)
print(sales)
print(sales_copy)
sales[-1].append(72)
print(sales)
print(sales_copy)

# Method 3: deep copy 
# Here is the way to create two completely independent copy of the object without persisting relationship in any level. It uses the module copy from python library. 
# deepcopy method will work for the independent copy making 
# copy method work as like the shallow copy 

import copy 
sales = [21, 34 , 45 ,65, 76, [12,34,43,44] ]
sales_copy = copy.deepcopy(sales)
sales.insert(2, 25)
print(sales)
print(sales_copy)
sales[-1].append(72)
print(sales)
print(sales_copy)
```

```python
# Combine the list 

# combine lists into one list by using + operator 
school_friends = ["Veera" , "Anand" ,"Ranieve"]
college_friends = ["Muthu" , "Sarvan" ,"Murugan" , "Ariki"]
work_friends_1 = ["Arun" , "Santhosh" , "Prasanth"]
work_friends_2 = ["Rahul" ,"Dinesh"]
friends = school_friends + college_friends + work_friends_1
print(friends) # the order based on in which order we combine 
friends = school_friends + work_friends_1 + college_friends
print(friends)

# Extend the list 
work_friends_1.extend(work_friends_2) # work_friends_1 get extended by the elements of passed list 
print(work_friends_1)
print(work_friends_2)

[list1 , list2] = [["a","b","c"], [1,2,3]]
print(list1)
print(list2)

# If we are in a situation to combine two list but combine each elements with  corresponding positional elements in other list

# zip() -> which return the iterator 
new_list = list(zip(list1, list2))
print(new_list)

new_list = list(zip(list2 , list1))
print(new_list)

# ENUMERATE -> RETURN THE LIST OF TUPLES WHICH CONTAINS THE LIST VALUE AND ITS INDEX 
print(list(enumerate(friends)))
```
